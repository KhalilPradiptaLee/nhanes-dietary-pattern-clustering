```python
import pandas as pd
```

```python
url1 = "https://drive.google.com/uc?export=download&id=15aoiF0m-IMe8PTXA-TrVVPYM8eLLq7sh"   # Dataset Demographic
url2 = "https://drive.google.com/uc?export=download&id=1Nn88QuFRIvVKOIPhcwXvIgkXW5tIGcjJ"   # Dataset Dairy Intake yang sudah digabung
url3 = "https://drive.google.com/uc?export=download&id=1ZguwctJserMqVzCuo7ZIVBWJDKKm05Dc"   # Dataset Gula Darah Puasa
url4 = "https://drive.google.com/uc?export=download&id=1kY8j0u2vCnP0bYzf9GA3lVyw8LnqIkmZ"   # Dataset Body Mass Index
url5 = "https://drive.google.com/uc?export=download&id=1BGHY7g1R6FK8fF1zUcn1tVoldi79OyDR"   # Dataset Blood Pressure
```

```python
# ==========================================
# 1. BACA SEMUA FILE DARI URL GOOGLE DRIVE
# ==========================================
demo = pd.read_csv(url1)
food = pd.read_csv(url2)
glu = pd.read_csv(url3)
bmx = pd.read_csv(url4)
bpx = pd.read_csv(url5)
```

```python
# ==========================================
# 2. PILIH KOLOM YANG DIPERLUKAN
# ==========================================

# Demographics
demo = demo[
    [
        "SEQN",
        "RIDAGEYR",
        "RIAGENDR"
    ]
]


# Dietary Food Group
food = food[
    [
        "SEQN",
        "milk_dairy",
        "meat_poultry_fish",
        "eggs",
        "legumes_nuts_seeds",
        "grain_products",
        "fruits_and_juices",
        "vegetables",
        "fats_oils_dressings",
        "sugars_sweets_beverages"
    ]
]


# Nutrient Intakes
# nutrient = nutrient[
#     [
#         "SEQN",
#         "DR1TKCAL",
#         "DR1TPROT",
#         "DR1TCARB",
#         "DR1TSUGR",
#         "DR1TFIBE",
#         "DR1TTFAT",
#         "DR1TCHOL"
#     ]
# ]


# Examination - BMI
bmx = bmx[
    [
        "SEQN",
        "BMXBMI"
    ]
].rename(
    columns={
        "BMXBMI": "BMI"
    }
)


# Examination - Blood Pressure
bpx = bpx[
    [
        "SEQN",
        "BPXSY1",
        "BPXSY2",
        "BPXSY3",
        "BPXDI1",
        "BPXDI2",
        "BPXDI3"
    ]
]
# .rename(
#     columns={
#         "BPXOSY1": "BPXSY1",
#         "BPXOSY2": "BPXSY2",
#         "BPXOSY3": "BPXSY3",
#         "BPXODI1": "BPXDI1",
#         "BPXODI2": "BPXDI2",
#         "BPXODI3": "BPXDI3"
#     }
# )


# Laboratory
glu = glu[
    [
        "SEQN",
        "LBDGLUSI"
    ]
]


# ==========================================
# 3. SAMAKAN TIPE DATA SEQN
# ==========================================

datasets = [
    demo,
    food,
    bmx,
    bpx,
    glu
]

for df in datasets:
    df["SEQN"] = pd.to_numeric(
        df["SEQN"],
        errors="coerce"
    ).astype("Int64")


# ==========================================
# 4. GABUNGKAN BERDASARKAN SEQN
# ==========================================

final_df = demo.copy()

final_df = final_df.merge(
    food,
    on="SEQN",
    how="left"
)

# final_df = final_df.merge(
#     nutrient,
#     on="SEQN",
#     how="left"
# )

final_df = final_df.merge(
    bmx,
    on="SEQN",
    how="left"
)

final_df = final_df.merge(
    bpx,
    on="SEQN",
    how="left"
)

final_df = final_df.merge(
    glu,
    on="SEQN",
    how="left"
)


# ==========================================
# 5. URUTAN KOLOM
# ==========================================

columns_order = [

    # Demographics
    "SEQN",
    "RIDAGEYR",
    "RIAGENDR",

    # Dietary Food Group
    "milk_dairy",
    "meat_poultry_fish",
    "eggs",
    "legumes_nuts_seeds",
    "grain_products",
    "fruits_and_juices",
    "vegetables",
    "fats_oils_dressings",
    "sugars_sweets_beverages",

    # Examination
    "BMI",
    "BPXSY1",
    "BPXSY2",
    "BPXSY3",
    "BPXDI1",
    "BPXDI2",
    "BPXDI3",

    # Laboratory
    "LBDGLUSI"
]

final_df = final_df[columns_order]


# ==========================================
# 6. CEK HASIL
# ==========================================

print("Ukuran dataset:", final_df.shape)
print("Jumlah SEQN unik:", final_df["SEQN"].nunique())

print("\nKolom:")
print(final_df.columns.tolist())

display(final_df.head())


# ==========================================
# 7. SIMPAN CSV
# ==========================================

output_file = "NHANES_combined_dataset.csv"

final_df.to_csv(
    output_file,
    index=False
)

print("\nDataset berhasil disimpan sebagai:")
print(output_file)
```
