# **Proses mengubah semua file .xpt menjadi .csv pada dataset**

```python
import pandas as pd
import gdown
from pathlib import Path
```

```python
urls = {
    "DR1IFF_J.xpt": "https://drive.google.com/uc?export=download&id=1iJuxkqvnYV9vMUOp-_crulqKqdG6-nE2",  # DR1IFF_J.xpt
    "DR2IFF_J.xpt": "https://drive.google.com/uc?export=download&id=1BOyFZJ7pKh-XFdGCN_ilwhIbdOFv1GH3",  # DR2IFF_J.xpt
    "DEMO_J.xpt"  : "https://drive.google.com/uc?export=download&id=1tfq9gPxGsjIfoL32EUEJmX5_lCsKcARh",  # DEMO_J.xpt
    "BMX_J.xpt"   : "https://drive.google.com/uc?export=download&id=1gkZqfk0LyFY85SimMWyMdsg7ibc0Ii8i",  # BMX_J.xpt
    "GLU_J.xpt"   : "https://drive.google.com/uc?export=download&id=1rW2r2iBzJrifgSyuPwiFlrr0aFcFqqp-",  # GLU_J.xpt
    "BPX_J.xpt"   : "https://drive.google.com/uc?export=download&id=1gpEKHrXp-vRg6jnJWaQ0Itj6FJvkV-M1"   # BPX_J.xpt
}
```

```python
input_folder = Path("xpt")
output_folder = Path("csv")

input_folder.mkdir(exist_ok=True)
output_folder.mkdir(exist_ok=True)
```

```python
print("=== DOWNLOAD FILE XPT ===\n")

for filename, url in urls.items():

    output_path = input_folder / filename

    print(f"Downloading: {filename}")

    result = gdown.download(
        url,
        str(output_path),
        quiet=False
    )

    if result is None:
        print(f"Download gagal: {filename}")
    else:
        print(f"Berhasil: {filename}")

    print("-" * 50)
```

```python
print("\n=== KONVERSI XPT MENJADI CSV ===\n")

xpt_files = list(input_folder.glob("*.xpt"))

for xpt_file in xpt_files:

    print(f"Sedang memproses: {xpt_file.name}")

    try:
        # Baca file XPT
        df = pd.read_sas(
            xpt_file,
            format="xport"
        )

        # Simpan menjadi CSV
        csv_file = output_folder / f"{xpt_file.stem}.csv"

        df.to_csv(
            csv_file,
            index=False
        )

        print(f"Berhasil: {csv_file}")
        print(f"Ukuran data: {df.shape}")

    except Exception as e:
        print(f"Gagal memproses {xpt_file.name}")
        print(f"Error: {e}")

    print("-" * 50)

print("\n=== PROSES SELESAI ===")

print("Jumlah file XPT:", len(xpt_files))
print("Jumlah file CSV:", len(list(output_folder.glob("*.csv"))))

print("\nLokasi hasil konversi:", output_folder)
```
